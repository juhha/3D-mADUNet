```python
import os, json, numpy as np
```

# Whole Brain Tumor Segmentation (BraTS 2021)

```python
print('BraTS 2021')
performance = json.load(open('../results/dice/brats_t2toflair/attdenseunet.json', 'r'))
mean = np.mean(performance)
std = np.std(performance)
print(f'\t{mean:.3f}$\pm${std:.2f}')
```

# Multi-organ Segmentation

```python
exp_names = ['synth_t1toct', 'synth_cbcttoct']
for exp_name in exp_names:
    print(exp_name)
    performance = json.load(open(f'../results/dice/{exp_name}/attdenseunet.json', 'r'))
    list_scores = [np.mean(d) for d in performance] # calculate the averagedice for each subject
    mean = np.mean(list_scores)
    std = np.std(list_scores)
    print(f'\t{mean:.3f}$\pm${std:.2f}')
```
