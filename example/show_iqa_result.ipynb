```python
import os, json, numpy as np
```

```python
exp_names = ['hcp1200_t1tot2', 'dhcp_t2tot1', 'brats_t2toflair', 'synth_t1toct', 'synth_cbcttoct']
for exp_name in exp_names:
    print(exp_name)
    performance = json.load(open(f'../results/iqa/{exp_name}/attdenseunet.json', 'r'))
    for metric_name in ['ssim', 'psnr', 'nmse']:
        print('\t', metric_name, f"{np.mean([performance[case_id][metric_name] for case_id in performance.keys()]):.3f}$\pm${np.std([performance[case_id][metric_name] for case_id in performance.keys()]):.2f}")    
```
